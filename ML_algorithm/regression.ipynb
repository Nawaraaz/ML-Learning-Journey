```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import PolynomialFeatures
from sklearn.metrics import mean_squared_error

# sample dataset for linear and polynomial regression
X = np.array([1,2,3,4,5,6]).reshape(-1,1)
y = np.array([1.5,3.8,6.7,12.5,20.3,30.1])

# Linear Regression
lin_model = LinearRegression()
lin_model.fit(X,y)
y_pred_lin = lin_model.predict(X)

# Polynomial Regression pf degree 2
poly = PolynomialFeatures(degree=2)
X_poly = poly.fit_transform(X)

poly_model = LinearRegression()
poly_model.fit(X_poly,y)
y_pred_poly = poly_model.predict(X_poly)

# Plotting results of linear and polynomial regression
plt.scatter(X,y)
plt.plot(X,y_pred_lin)
plt.plot(X,y_pred_poly)
plt.title("Linear vs Polynomial Regression")
plt.show()

print("Linear MSE:",mean_squared_error(y,y_pred_lin))
print("Polynomial MSE:",mean_squared_error(y,y_pred_poly))
```
