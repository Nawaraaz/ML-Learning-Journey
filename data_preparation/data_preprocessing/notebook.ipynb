```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler

data = {
    "Age": [22, 25, np.nan, 28, 35, np.nan],
    "Salary": [25000, 32000, 28000, np.nan, 52000, 48000],
    "Department": ["IT", "HR", "IT", "Finance", np.nan, "HR"],
    "Gender": ["Male", "Female", "Female", "Male", "Female", np.nan]
}
df = pd.DataFrame(data)
print("Original Dataset:\n")
print(df)

# 2. Handle Missing Values

df["Age"].fillna(df["Age"].mean(), inplace=True)
df["Salary"].fillna(df["Salary"].mean(), inplace=True)

df["Department"].fillna(df["Department"].mode()[0], inplace=True)
df["Gender"].fillna(df["Gender"].mode()[0], inplace=True)

# 3. Encode Categorical Features

df_encoded = pd.get_dummies(df, columns=["Department", "Gender"])

# 4. Scale Numerical Features

scaler = StandardScaler()
df_encoded[["Age", "Salary"]] = scaler.fit_transform(df_encoded[["Age", "Salary"]])

# 5. Final Preprocessed Dataset

print("\nPreprocessed Dataset:\n")
print(df_encoded)
```
